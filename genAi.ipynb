```python
with open("the-verdict.txt","r") as f:
  raw_text = f.read()
```

```python
import re
result = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

```python

all_words = sorted(result)
vocab = list(set(all_words))
vocab.extend(["<|endoftext|>","<|unk|>"]) #added special tokens
print(vocab[-5:])
```

```python
dict = {}
for i in range(len(vocab)):
  dict[vocab[i]] = i
```

```python
print(list(dict.items())[-5:])
```

```python
class Tokenizer:
  def __init__(self,dict):
    self.str_to_int = dict
    self.int_to_str = {i:s for s,i in dict.items()}
  def encode(self,raw_text):
    result = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
    result = [item.strip() for item in result if item.strip()]
    result = [  item if item in self.str_to_int
                else "<|unk|>" for item in result
              ]
    ids = [self.str_to_int[s] for s in result]
    return ids
  def decode(self,ids):
    result = " ".join([self.int_to_str[i] for i in ids])
    result = re.sub(r'\s+([,.?!"()\'])', r'\1', result)
    return result
```

```python
tokenizer = Tokenizer(dict)
text1 = "Hello,do you like cake?"
text2 = "humanity is in the verge of extention."
text = "<|endoftext|>".join((text1,text2))
print(text)
```

```python
tokenizer.encode(text)
```

```python
tokenizer.decode(tokenizer.encode(text))
```
